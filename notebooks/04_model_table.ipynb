# 04 Model table

Builds `data/processed/model_table.csv`: one row per driver per race, 2014-2026.
Kaggle supplies results and lap times up to 2024; Jolpica-F1 supplies 2025 and 2026
(downloaded with `src/fetch_jolpica.py`). All the logic is in `src/data_prep.py`.

```python
import sys
from pathlib import Path

# find the repo root (the folder that contains src/) wherever the notebook is started from
cwd = Path.cwd()
root = next(p for p in [cwd, cwd / "AdvanceData", *cwd.parents] if (p / "src").is_dir())
sys.path.insert(0, str(root / "src"))

import data_prep
```

```python
table = data_prep.build_model_table()
data_prep.save_model_table(table)

features = ["grid", "team_pace3", "driver_form3", "driver_wins_before",
            "driver_team_wins_before", "driver_races_before"]

sel = table[table["year"].between(2014, 2024)]
print("rows 2014-2024:", len(sel), "(expected 4626)")
print("wins 2014-2024:", sel["win"].sum(), "(expected 228, one per race)")
print("missing values 2014-2024:")
print(sel[features].isna().sum())

recent = table[table["year"] >= 2025]
print("missing team_pace3 in 2025-2026:")
print(recent.groupby("year")["team_pace3"].apply(lambda s: s.isna().sum()))

modern = table[table["year"] >= 2014]
print("races without exactly one winner:", (modern.groupby("race_id")["win"].sum() != 1).sum())
```
